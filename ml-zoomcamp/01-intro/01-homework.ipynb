```python
# median value of horsepower
median_value = df['horsepower'].median()
median_value
```

```python
# most frequent value in horsepower column
freq_horsepower = df['horsepower'].mode()[0]
freq_horsepower
```

```python
# fill missing horsepower with most freq horsepower
df['horsepower'] = df['horsepower'].fillna(freq_horsepower)
```

```python
# new median
new_median_value = df['horsepower'].median()
new_median_value
```

```python
if new_median_value > median_value:
    print('increased')
elif new_median_value < median_value:
    print('decreased')
else:
    print('no change')
```

```python
# select all asian cars
asian_cars = df[df['origin'] == 'Asia']
asian_cars.head()
```

```python
# select only vehicle_weight and model_year
new_df = asian_cars[['vehicle_weight','model_year']]
new_df
```

```python
# select first 7 values
new_df = new_df.head(n=7)
new_df
```

```python
# get underlying numpy array
X = new_df.values
X
```

```python
# X transpose
XT = X.T
XT
```

```python
# matrix-matrix multiplication XT and X
XTX = XT.dot(X)
XTX
```

```python
# invert XTX
invert_XTX = np.linalg.inv(XTX)
invert_XTX
```

```python
# new array y with given values
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

```python
# multiply inverser of XTX and transpose of X
w = (invert_XTX.dot(XT)).dot(y)
w
```

```python
# sum of all elements
w.sum()
```
